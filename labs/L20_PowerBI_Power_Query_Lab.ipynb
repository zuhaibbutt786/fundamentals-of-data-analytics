# L20 Lab: Power BI — Data Import & Power Query

**Tool:** Power BI Desktop  
**Goal:** Connect to CSV, shape data in Power Query, load a clean table.

## Dataset
- `datasets/sample_sales.csv` or `sample_accidents.csv`

## Steps

### 1. Get Data
1. **Get data** → **Text/CSV** → select file → **Transform Data**.

### 2. Profile
1. View → enable **Column quality**, **distribution**, **profile**.
2. Note errors, empties, high cardinality.

### 3. Transforms (do at least four)
1. Promote headers / set types explicitly.
2. Replace values or errors.
3. Filter nulls on a key column.
4. Split or merge columns if needed.
5. Add conditional column (e.g. High/Low flag).
6. Optional: Group By sum by region.

### 4. Close & Apply
Create one table visual and one card (e.g. Sum of Revenue).
