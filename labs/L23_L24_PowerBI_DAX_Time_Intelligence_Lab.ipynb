# L23–L24 Lab: Power BI — DAX Basics & Time Intelligence

**Tool:** Power BI Desktop (model from L22 or simple single-table model)  
**Goal:** Core measures + YTD / Prior Year / YoY %.

## Part A — Core measures (L23)

```dax
Total Sales := SUM ( FactSales[Revenue] )

Order Count := COUNTROWS ( FactSales )

Average Order Value := DIVIDE ( [Total Sales], [Order Count] )

Sales All Regions :=
CALCULATE ( [Total Sales], REMOVEFILTERS ( DimRegion[Region] ) )

Sales % of All :=
DIVIDE ( [Total Sales], [Sales All Regions] )
```

Test with Region on rows in a matrix.

## Part B — Time intelligence (L24)

**Requires:** marked continuous date table related to the fact.

```dax
Sales YTD := TOTALYTD ( [Total Sales], DimDate[Date] )

Sales Prior Year :=
CALCULATE ( [Total Sales], SAMEPERIODLASTYEAR ( DimDate[Date] ) )

Sales YoY % :=
DIVIDE ( [Total Sales] - [Sales Prior Year], [Sales Prior Year] )

Sales Prior Month :=
CALCULATE ( [Total Sales], DATEADD ( DimDate[Date], -1, MONTH ) )
```

## Part C — Running total (stretch)

```dax
Sales Running Total :=
CALCULATE (
    [Total Sales],
    FILTER ( ALL ( DimDate[Date] ), DimDate[Date] <= MAX ( DimDate[Date] ) )
)
```
