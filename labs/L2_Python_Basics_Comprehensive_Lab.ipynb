```python
records = [10, 20, 30, 40]
record = {"quantity": 3, "unit_price": 200}
def revenue(quantity, price):
    return quantity * price
print(records[1:3], revenue(record["quantity"], record["unit_price"]))
for value in records:
    if value >= 30:
        print("large", value)
assert revenue(3, 200) == 600
```

## Student task
Write a function that returns net revenue after a fractional discount.

## Reference solution
Try the task before reading this cell.

```python
def net_revenue(quantity, price, discount):
    assert 0 <= discount <= 1
    return quantity * price * (1-discount)
assert net_revenue(3,200,.1) == 540
print(net_revenue(3,200,.1))
```
