## 1. Variables and Data Types
Variables are containers for storing data values.

```python
# Integer
age = 25
# Float
price = 19.99
# String
name = "Data Science"
# Boolean
is_valid = True

print(f"Type of age: {type(age)}")
print(f"Type of price: {type(price)}")
```

## 2. String Manipulation
Working with text data often requires splitting or cleaning strings.

```python
raw_data = "apple,banana,cherry,date"

# Splitting a string into a list
fruit_list = raw_data.split(",")
print("Split List:", fruit_list)

# Joining a list into a string
joined_string = " - ".join(fruit_list)
print("Joined String:", joined_string)

# Stripping whitespace
dirty_string = "   clean me   "
print("Stripped:", dirty_string.strip())
```

## 3. Indexing & Slicing
Python uses 0-based indexing. 
Formula: `[start : stop : step]` (Stop is exclusive).

```python
numbers = [0, 10, 20, 30, 40, 50]
print("First element:", numbers[0])
print("Last element:", numbers[-1])
print("Slicing 10 to 30:", numbers[1:4])
print("Every second number:", numbers[::2])
print("Reverse list:", numbers[::-1])
```

## 4. Conditional Statements
Logic to control the flow of the program.

```python
score = 85

if score >= 90:
    print("Grade: A")
elif score >= 80:
    print("Grade: B")
else:
    print("Grade: C")
```

## 5. Loops
* **For Loops**: Iterate over a sequence (list, string, range).
* **While Loops**: Repeat while a condition is True.

```python
print("For Loop:")
for i in range(3):
    print(f"Iteration {i}")

print("\nWhile Loop:")
count = 0
while count < 3:
    print(f"Count {count}")
    count += 1
```
