## Part 1: Setup — synthetic e-commerce context
Imagine a year of online transactions, marketing campaigns, and support tickets.

```python
import pandas as pd
import numpy as np

# Tiny synthetic sample (replace with course sample_sales.csv when available)
np.random.seed(42)
n = 100
df = pd.DataFrame({
    "order_id": range(1, n+1),
    "region": np.random.choice(["North", "South", "East", "West"], n),
    "channel": np.random.choice(["Web", "Mobile", "Store"], n),
    "revenue": np.random.uniform(20, 500, n).round(2),
    "returned": np.random.choice([0, 1], n, p=[0.9, 0.1])
})
print(df.head())
print(df.shape)
```

### Challenge 1: "Why is our revenue soft this quarter?"

### Challenge 3: "Are returns concentrated in one region?"
Propose a grouped analysis and a simple visual.

```python
# Example starter for Challenge 3
summary = df.groupby("region").agg(
    orders=("order_id", "count"),
    revenue=("revenue", "sum"),
    return_rate=("returned", "mean")
).round(3)
print(summary)
```
