# L11 Lab: Descriptive Statistics

**Goal:** Center, spread, shape; choose the right summary.

**Dataset:** `../datasets/sample_sales.csv` or `pima_diabetes.csv`


```python
import pandas as pd
import numpy as np

df = pd.read_csv("../datasets/sample_sales.csv")
num = df.select_dtypes("number")
print(num.describe().T)
```

```python
col = num.columns[0]
s = df[col].dropna()
print("Mean:", s.mean())
print("Median:", s.median())
print("Std / IQR:", s.std(), s.quantile(0.75) - s.quantile(0.25))
print("Skew / Kurtosis:", s.skew(), s.kurtosis())
```
