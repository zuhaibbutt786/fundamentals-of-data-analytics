# L14 Lab: Time-Series Exploration

**Goal:** Parse dates, moving averages, trend insights.

**Dataset:** `../datasets/electric_production.csv`


```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("../datasets/electric_production.csv")
print(df.head())
print(df.columns.tolist())
```

```python
date_col = [c for c in df.columns if any(k in c.lower() for k in ("date", "month", "time"))]
val_col = [c for c in df.columns if c not in date_col][0]
if date_col:
    df[date_col[0]] = pd.to_datetime(df[date_col[0]])
    df = df.sort_values(date_col[0]).set_index(date_col[0])
series = df[val_col].astype(float)
ma_short = series.rolling(7, min_periods=1).mean()
ma_long = series.rolling(30, min_periods=1).mean() if len(series) > 30 else series.rolling(5, min_periods=1).mean()
plt.figure(figsize=(10, 4))
series.plot(label="Raw", alpha=0.5)
ma_short.plot(label="MA short")
ma_long.plot(label="MA long")
plt.legend(); plt.title("Time series with moving averages"); plt.tight_layout(); plt.show()
```
