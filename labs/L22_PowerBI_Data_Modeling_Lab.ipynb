# L22 Lab: Power BI — Star Schema Data Modeling

**Tool:** Power BI Desktop  
**Goal:** Fact + dimensions, relationships, marked date table.

## Setup
Split `sample_sales.csv` in Power Query into FactSales + DimRegion/DimProduct + DimDate, **or** use two related CSVs.

## Steps

### 1. Grain
Write: *One row in the fact table = …*

### 2. Dimensions
1. **DimDate** from date column (Year, Month, MonthName, Quarter).
2. At least one other dimension (Product or Region) with key + attributes.

### 3. Model view
1. Fact → each dimension: **many-to-one**, single direction.
2. **Mark DimDate as date table**.
3. Hide foreign keys on the fact from Report view.

### 4. Validate
Matrix: dimension on rows, sum measure on values. Compare to an Excel pivot.

## Deliverable
- [ ] Grain documented
- [ ] ≥ 2 relationships, correct cardinality
- [ ] Date table marked
- [ ] Model view screenshot
