## 1. Theory: Data Structures

| Type | Description | Examples |
| :--- | :--- | :--- |
| **Structured** | Highly organised, fixed schema | Relational tables, CSV with clear columns |
| **Semi-structured** | Tags or keys, flexible schema | JSON, XML, NoSQL documents |
| **Unstructured** | No predefined model | Text, images, audio, video |

## 2. Ethics & Legalities
Before scraping, always check:
1. **robots.txt** of the target site
2. Terms of Service
3. Whether personal data is involved (privacy law)
4. Rate limits — do not overload servers

Prefer official APIs and published open data when available.
