# L7 Lab: Data Transformation

**Goal:** Apply scaling, encoding, and simple transforms.

**Dataset:** `../datasets/sample_sales.csv`


```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler, MinMaxScaler

df = pd.read_csv("../datasets/sample_sales.csv")
print(df.dtypes)
print(df.head())
```

## Standardization vs Min-Max
- **StandardScaler**: mean 0, std 1
- **MinMaxScaler**: [0, 1] range


```python
num_cols = [c for c in df.select_dtypes("number").columns if "id" not in c.lower()][:2]
print("Numeric cols:", num_cols)
scaler = StandardScaler()
df_std = df.copy()
df_std[num_cols] = scaler.fit_transform(df[num_cols])
print(df_std[num_cols].describe().round(3))
```

## Encoding categoricals


```python
cat_cols = df.select_dtypes("object").columns.tolist()[:2]
print("Categorical:", cat_cols)
if cat_cols:
    dummies = pd.get_dummies(df[cat_cols], drop_first=True)
    print(dummies.head())
```
